```python
# Remove missing values
skills_series = df["Skills"].dropna()

# Combine all skills
all_skills = " ".join(skills_series)

# Split into list
skills_list = all_skills.split()

# Count skills
top_skills = Counter(skills_list).most_common(20)

# Convert to DataFrame
skills_df = pd.DataFrame(
    top_skills,
    columns=["Skill", "Count"]
)

print(skills_df)
```

```python
import matplotlib.pyplot as plt

skills_df.plot(
    x="Skill",
    y="Count",
    kind="bar",
    figsize=(12,6)
)

plt.title("Top 20 Most Demanded Skills")

plt.xlabel("Skills")

plt.ylabel("Frequency")

plt.xticks(rotation=45)

plt.show()
```
